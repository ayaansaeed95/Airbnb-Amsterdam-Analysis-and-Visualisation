Handling Misssing Values

```python
#Check count of unique values for bathroom_text

df['bathrooms_text'].value_counts()
```

```python
#Fill most common value
#There are only 15 rows missing this columnn so filling most common value should not greatly disturb analysis

df['bathrooms_text'] = df['bathrooms_text'].fillna('1 bath')
```

```python
#Fill missing night-limit columns with the median value of that column, which is standard imputation strategy for handling missing values

min_max_cols = [
    'minimum_minimum_nights', 'maximum_minimum_nights',
    'minimum_maximum_nights', 'maximum_maximum_nights',
    'minimum_nights_avg_ntm', 'maximum_nights_avg_ntm'
]

for col in min_max_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
#Handling review columns with large missing values since they have >500 missing values
review_columns = ['first_review', 'last_review', 
    'review_scores_rating', 'review_scores_value',
    'reviews_per_month']

# Flagging rows with missing review values
df['missing_review_info'] = df['review_scores_rating'].isnull().astype(int)

#Converting str dates to date_time
df['first_review'] = pd.to_datetime(df['first_review'])
df['last_review'] = pd.to_datetime(df['last_review'])

#For missing values of review_scores_rating, imputating the missing values by adding the median value of the type of room it is
df['review_scores_rating'] = df['review_scores_rating'].fillna(
    df.groupby('room_type')['review_scores_rating'].transform('median')
)

#Do same for review_scores_value
df['review_scores_value'] = df['review_scores_value'].fillna(
    df.groupby('room_type')['review_scores_value'].transform('median')
)

#adding 0 in the case there are no reviews
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

#Some others are categorical variables and have
# quite a few null values
#so it is not appropriate to assign them all to the same 
#response. Thus, we fill them with the same distribution 
# as the exisitng rows, this way we preserve the 
#distribution of the existing values, 
#and not make extra assumptions about the data

cols = ['host_response_time', 'bedrooms', 'beds']

for col in cols:
    missing_idx = df[col].isnull()
    df.loc[missing_idx, col] = (
        df[col]
        .dropna()   
        .sample(missing_idx.sum(), replace=True)
        .values
)
```

```python
# Verifying that listings without first and last reviews Have 0 in number_of_reviews
print(df[df['first_review'].isnull()]['number_of_reviews'].unique())
```

```python
#creating a column to show whether there are reviews or not
df['has_reviews'] = df['first_review'].notnull().astype(int)

#To fill missing values for first and last review, we fill 
#in dates from host_since and last_scraped, making the
#assumption that if there were reviews, they were close to
#dates of registration and last scraping respectively

#first converting str to datetimes
df['last_scraped'] = pd.to_datetime(df['last_scraped'])
df['last_review'] = pd.to_datetime(df['last_review'])
#then imputing missing values
df['first_review'] = df['first_review'].fillna(df['host_since'])
df['last_review'] = df['last_review'].fillna(df['last_scraped'])
```

```python
# create a new column named review_status
# If the listing has 0 reviews then it's labeled 'No Reviews'
# Creating days_since_last_review Column
# to calculate the number of days since the last review
df['review_status'] = np.where(df['number_of_reviews'] == 0, 'No Reviews', 'Has Reviews')
df['days_since_last_review'] = (df['last_scraped'] - df['last_review']).dt.days
```
